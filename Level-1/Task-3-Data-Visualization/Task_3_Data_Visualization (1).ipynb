```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/2) Stock Prices Data Set.csv')

df['date'] = pd.to_datetime(df['date'])
print(df.head())
```

```python
aapl = df[df['symbol'] == 'AAPL']

plt.figure(figsize=(12,6))
plt.plot(aapl['date'], aapl['close'])
plt.title('AAPL Closing Price Trend')
plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top5 = df.groupby('symbol')['close'].mean().sort_values(ascending=False).head(5)

plt.figure(figsize=(8,5))
top5.plot(kind='bar')
plt.title('Top 5 Stocks by Average Closing Price')
plt.xlabel('Company Symbol')
plt.ylabel('Average Close Price')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df.sample(1000),
    x='open',
    y='close'
)

plt.title('Open Price vs Close Price')
plt.xlabel('Open Price')
plt.ylabel('Close Price')
plt.show()
```
